```python
# this needs to be adjusted to the updates path
deltaTable_target = DeltaTable.forPath(spark, data['destination_table']['path'])

# this needs to be adjusted to the updates path --> check how this works with cdf updates
deltaTable_source = DeltaTable.forPath(spark, data['source_table']['path'])

dfUpdates = deltaTable_source.toDF()
```

```python

deltaTable_target.alias('target') \
  .merge(
    dfUpdates.alias('source'),
    'source.AddressID = target.AddressID'
  ) \
  .whenMatchedUpdate(set =
    {
      "AddressID": "source.AddressID",
      "AddressLine1": "source.AddressLine1",
      "AddressLine2": "source.AddressLine2",
      "City": "source.City",
      "StateProvince": "source.StateProvince",
      "CountryRegion": "source.CountryRegion",
      "PostalCode": "source.PostalCode",
      "rowguid": "source.rowguid"
    }
  ) \
  .whenNotMatchedInsert(values =
    {
      "AddressID": "source.AddressID",
      "AddressLine1": "source.AddressLine1",
      "AddressLine2": "source.AddressLine2",
      "City": "source.City",
      "StateProvince": "source.StateProvince",
      "CountryRegion": "source.CountryRegion",
      "PostalCode": "source.PostalCode",
      "rowguid": "source.rowguid"
    }
  ) \
  .execute()
```
